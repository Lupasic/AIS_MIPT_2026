## 1 Hello World!
- Gradio is a library for creating web interfaces for applications.
- Calling `launch()` starts a local web server that accepts HTTP requests from users.

```python
import gradio as gr

def greet(name):
    return f"Hello, {name}! "

demo = gr.Interface(fn=greet, inputs=["text"], outputs="text")
demo.launch(share=True, debug=True)
```

## 2 Core UI Components
- Text and numbers:
  - `gr.Textbox()`: multi-line text input
  - `gr.Number()`: numeric input
  - `gr.Slider()`: select a number from a range

- Category selection:
  - `gr.Checkbox()`: boolean flag (input, output)
  - `gr.Radio(choices=...)` or `gr.CheckboxGroup(choices=...)`: select one or several categories from a list
  - `gr.Dropdown(choices=..., multiselect=...)`: select from a dropdown list.

- Buttons:
  - `gr.Button("Label")`: a button.
  - `gr.ClearButton()`: a special button to reset a component's value.

- Media:
  - `gr.Image()`: upload and display images
  - `gr.File()`: upload and download files.
  - `gr.Audio()`, `gr.Video()`: audio and video files.

- Miscellaneous:
  - `gr.Dataframe()`: interactive table
  - `gr.Label()`: display predicted class
  - `gr.HTML()`, `gr.Markdown()`: render HTML and Markdown
  - `gr.ColorPicker()`, `gr.Datepicker()`, etc.

```python
import gradio as gr

def greet(name, excitement_level):
    return f"Hello, {name}" + "!" * int(excitement_level)

demo = gr.Interface(
    fn=greet,
    inputs=[gr.Textbox(label="Your Name"), gr.Slider(minimum=1, maximum=5, value=2, label="Excitement Level")],
    outputs=gr.Textbox(label="Greeting"),
    title="Greeter App",
    description="Enter your name and choose an excitement level."
)
demo.launch(share=True, debug=True)
```

### 2.1 Calculator

```python
def calculator(a, operation, b):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        return None if b == 0 else a / b

demo = gr.Interface(
    fn=calculator,
    inputs=[
        gr.Number(label="First Number"),
        gr.Radio(["add", "subtract", "multiply", "divide"], label="Operation"),
        gr.Number(label="Second Number")
    ],
    outputs=gr.Textbox(label="Result"),
    title="Simple Calculator"
)
demo.launch(share=True, debug=True)
```

## 3 Blocks

For more complex interfaces, use the Blocks API. `gr.Blocks()` allows using context managers to lay out components in rows/columns and bind events (e.g., button clicks, value changes) to functions.

### 3.1 Tracking Field Value Changes

In the example, moving the slider automatically updates the text field: `age_input.change(...)` binds the slider to the `age_alert` function, which returns a new string.

```python
import gradio as gr

def age_alert(age):
    return f"You are {age} years old!"

with gr.Blocks() as demo:
    age_input = gr.Slider(minimum=0, maximum=100, label="Select your age")
    output = gr.Textbox(label="Age Info")
    age_input.change(fn=age_alert, inputs=age_input, outputs=output)

demo.launch(share=True, debug=True)
```

### 3.2 Triggering an Action on Button Click

In the example, clicking the button resets the value to the default.

```python
import gradio as gr

def age_alert(age):
    return f"You are {age} years old!"

with gr.Blocks() as demo:
    age_input = gr.Slider(minimum=0, maximum=100, value=50, label="Select your age")
    output = gr.Textbox(label="Age Info")
    button = gr.Button("Default Value")
    button.click(fn=lambda: 50, inputs=None, outputs=age_input)
    age_input.change(fn=age_alert, inputs=age_input, outputs=output)

demo.launch(share=True, debug=True)
```

## 4 Model Integration

General pattern: input components (Text, Slider, ...) pass arguments to a Python function that uses a model/API, then output components display the results.

### 4.1 Text Sentiment Analysis

```python
from transformers import pipeline
import gradio as gr

# Load a pretrained sentiment-analysis pipeline
sentiment_pipeline = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")

def analyze_sentiment(text):
    result = sentiment_pipeline(text)[0]
    label = result['label']
    score = result['score']
    return f"Sentiment: {label}, confidence {score:.2f}"

gr.Interface(
    fn=analyze_sentiment,
    inputs=gr.Textbox(lines=3, placeholder="Type a sentence...", label="Input Text"),
    outputs=gr.Textbox(label="Sentiment"),
    title="Sentiment Analyzer",
    description="Enter text to analyze its sentiment."
).launch(share=True, debug=True)
```

### 4.2 Image Classification

```python
from transformers import pipeline
import gradio as gr

img_classifier = pipeline("image-classification")

def classify_image(img):
    results = img_classifier(img)
    top3 = results[:3]
    formatted_results = "\n".join([f"{res['label']}: {res['score']*100:.2f}%" for res in top3])
    return formatted_results

demo_image = gr.Interface(
    fn=classify_image,
    inputs=gr.Image(type="pil"),
    outputs=gr.Textbox(lines=3, label="Result"),
    title="Image Classification Demo",
    description="Upload an image and see the top predicted labels from a pre-trained model.",
    examples=[
        ["https://raw.githubusercontent.com/gradio-app/gradio/main/test/test_files/bus.png"],
    ]
)

demo_image.launch()
```
